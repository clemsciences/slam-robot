 # Un peu de géométrie pour se repérer
 
On va revoir quelques définitions et quelques propriétés concernant la géométrie dans le plan (en dimension 2). Plus précisément, on se place dans un espace affine euclidien de dimension 2. 

:::{note}
Un espace affine euclidien est la géométrie telle qu'on l'a apprise depuis le début de notre scolarité. Bien que les vecteurs soient abordés que bien plus tard, ils en sont une partie essentielle.

On se donne un espace vectoriel E sur le corps $\mathbb{K}$. On appelle un espace affine sur $\mathbb{K}$ d'espace directeur E, un ensemble $\mathcal{E}$ non vide muni d'une application $\phi$ de $\mathcal{E} \times \mathcal{E} \to E$ tel que $(x, y) \mapsto \overrightarrow{xy}$ ou $\phi(x, y) = \overrightarrow{xy} \in \mathnormal{E}$ avec les deux propriétés suivantes :
* $\forall\ (x, y, z) \in \mathcal{E}^3, \overrightarrow{xy} + \overrightarrow{yz} = \overrightarrow{xz}$
* $\forall\ x \in \mathcal{E}, \forall u \in E, \exists!\ y \in \mathcal{E} / \overrightarrow{xy} = u$
:::

Plus rigoureusement parlant, l'espace affine euclien est $\mathbb{R}^2$ sur le corps $\mathbb{R}$.

## Produit scalaire

Le produit scalaire est une opération qui agit sur deux vecteurs. Dans notre espace, il prend la forme suivante pour les vecteurs $u$ et $v$ : $u \cdot v = x_u x_v + y_u y_v$.

On définit la norme comme étant la racine carré du produit scalaire d'un vecteur par lui-même.

$||u|| = \sqrt{u \cdot u}$

Une propriété intéressante des produits scalaires est que le produit de deux vecteurs non nuls est nul si et seulement si les vecteurs sont orthogonaux.

## Distance

On peut aussi calculer la distance entre deux points $p_{1}(x_1, y_1)$ et $p_{2}(x_2,y_2)$ : $d(p_{1}, p_{2}) = \sqrt{(x_{1} - x_{2})^{2} + (y_{1} - y_{2})^{2}}$.

:::{note}
La distance entre deux points provient de la norme du vecteur qui relie les deux points. Soient $(A, B) \in \mathcal{E}$, on note ce vecteur $\overrightarrow{AB}$. On peut aussi écrire $B - A = \overrightarrow{AB}$ dont les composantes sont
$\begin{pmatrix}
x_{B} - x_{A} \\
 y_{B} - y_{A}
\end{pmatrix}$.
:::

```python
point_1 = Point(1, 2)
point_2 = Point(3, 4)
math.sqrt((point_1.x - point_2.x)**2 + (point_1.y - point_2.y)**2)
```

```python
point_1.distance(point_2)
```

## Droite
Une droite est la donnée d'un point et d'un vecteur directeur.

Soient A un point de E et $\overrightarrow{u}$ de $\mathcal{E}$ non nul. Un point M appartient à la droite D si et seulement s'il existe un réel t tel que  $M = A + t\ \overrightarrow{u}$. On peut réécrire cela $D = \{t \in \mathbb{R}: A + t\ \overrightarrow{u}\}$

:::{note}
Une droite peut être définie formellement comme un sous-espace affine de E de dimension 1. Cela signifie que l'espace vectoriel associé est de dimension 1 et donc qu'un seul vecteur constitue une base de celle-ci.

:::

Une équation paramétrique d'une droite est de la forme : $t \in \mathbb{R} \mapsto A + t\ \overrightarrow{u}$

Une équation cartésienne d'une droite est de la forme : $a x + b y + c = 0$, avec $a, b, c$ des coefficients réels et $(x, y)$ les coordonnées réelles d'un point de la droite.

## Cercle

On définit classiquement un cercle comme étant l'ensemble des points qui sont à une certaine distance d'un point.

Soit C, un cercle de centre $(x_{c}, y_{c})$ et de rayon r. Pout un point P de coordonnées $(x, y)$, on a :
$P \in C \iff (x - x_{c})^2 +(y - y_{c})^2 = r^2$
