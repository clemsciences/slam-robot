## Le point

On peut définir un point comme un élément de $\mathbb{R}^2$. Par exemple $p = (x, y)$ avec $x \in \mathbb{R}$ et $y \in \mathbb{R}$

```python
import math
from slam_robot.models.world_items import Point
point_1 = Point(10, 13)
```

```python
point_1.x, point_1.y
```

L'angle d'un point, c'est l'angle que fait la droite passant par l'origine (0, 0) et par le point en question et l'axe des abscisses (Ox).

```python
point_1.to_angle()
```

```python
vector = point_1.from_other_point_to_vector(Point(10, 15))
```

On peut aussi calculer la distance entre deux points $p_{1}(x_1, y_1)$ et $p_{2}(x_2,y_2)$ : $d(p_{1}, p_{2}) = \sqrt{(x_{1} - x_{2})^{2} + (y_{1} - y_{2})^{2}}$.

```python
point_1 = Point(1, 2)
point_2 = Point(3, 4)
math.sqrt((point_1.x-point_2.x)**2 + (point_1.y - point_2.y)**2)
```

```python
point_1.distance(point_2)
```

# Cercle

On définit classiquement un cercle comme étant l'ensemble des points qui sont à une certaine distance d'un point.

Soit C, un cercle de centre $(x_{c}, y_{c})$ et de rayon r. Pout un point P de coordonnées $(x, y)$, on a :
$P \in C \iff (x - x_{c})^2 +(y - y_{c})^2 = r^2$

```python
from slam_robot.models.world_items import Circle
```

```python
center = Point(10, 10)
radius = 5
circle = Circle(center, radius)
```
