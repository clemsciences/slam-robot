 # Construire l'environnement
 
On décrit d'abord l'environnement qui est une table sous forme d'un carré de 100 cm par 100 cm.

Qu'est-ce qui compte réellement dans l'élabroration de l'environnement de notre robot ? Ce sont les interactions que le robot peut avoir avec son environnement.
Quelles sont les formes que prennent ces interactions ? On parle ici d'interactions physiques (tout ce qui est métaphysique avec un robto est hors de propos ici).

Comme nous utilisons un LiDAR, on va s'intéresser à la propriété qu'ont les matériaux à refléter un rayon laser dans la direction d'émission du même rayon.

Et comme nous avons un robot, c'est-à-dire un agent mécanique capable d'agir dans le monde physique, nous devons prendre en compte les endroits dans lesquels il peut aller, mais aussi les endroits où il ne peut pas aller. A l'intersection de ces deux ensembles, nous avons les obstacles que l'on doit **contourner** pour éviter des **collisions**.
