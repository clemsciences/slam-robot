 # Construire l'environnement du robot

Qu'est-ce qui compte réellement dans l'élabroration de l'environnement de notre robot ? Ce sont les interactions que le robot peut avoir avec son environnement.
Quelles sont les formes que prennent ces interactions ? On parle ici d'interactions physiques (tout ce qui est métaphysique avec un robto est hors de propos ici).



Et comme nous avons un robot, c'est-à-dire un agent mécanique capable d'agir dans le monde physique, nous devons prendre en compte les endroits dans lesquels il peut aller, mais aussi les endroits où il ne peut pas aller. A l'intersection de ces deux ensembles, nous avons les obstacles que l'on doit **contourner** pour éviter des **collisions**.  

Il suffit de définir les obstacles et de considérer que tout ce qui n'est pas un obstacle est traversable par le robot.

On définit une liste d'obstacles.


```python
items = []
```

Nous pouvons ajouter différents obstacles que nous avons rencontrés au chapitre précédent (voir [géométrie](1_geometry.ipynb)).
On commence par une colonne représentée par un cercle.

```python
from slam_robot.models.world_items import Circle, Point

items.append(Circle(Point(20, 40), 15))
```

Puis un mur représenté par un segment.

```python
from slam_robot.models.world_items import LineByTwoPoints
items.append(LineByTwoPoints(Point(80, 80), Point(90, 100)))
```

Les obstacles circulaires et rectilignes sont les deux types d'obstacle que nous abordons pour le moment. Sauf à considérer un environnement ouvert, c'est-à-dire sans frontière, nous ajoutons aux contraintes de déplacement des limites aux bords. Choisissons des bords tels qu'ils soient rectangulaires. Par exemple (100, 100).

```python
from slam_robot.models.world import World
world = World(items, 100, 100)
```

```python
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (5, 5)
ax = plt.axes()
world.draw(ax)
```
