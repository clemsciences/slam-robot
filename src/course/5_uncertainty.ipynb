# L'incertitude ajoutée

Après avoir cotoyé la physique au collège et au lycée, nous sommes tentés de voir en la science la capcité de comprendre ce qui nous entoure avec certitude. N'avons-nous pas des lois qui sont tout le temps vérifiés ?

La réponse est oui, mais c'est insuffisant pour affirmer que les processus physiques sont forcément déterministes. Plusieurs raisons peuvent nous amener à envisager des facteurs aléatoires à ces processus. Il y a tout d'abord la méconnaissance des conditions initiales exactes. La physique quantique, avec son inégalité de Heisenberg, nous enseigne qu'il est impossible d'avoir une connaissance aussi poussée que voulue de plusieurs grandeurs physiques (on parle d'observables qui ne sont pas commutatives). De plus, par définition d'un système quantique, celui-ci est représenté par une densité de probabilité. Et nous ne pouvons accéder à l'état d'un système quantique qu'après une mesure, qui détruit l'état de superposition potentielle d'une particule. Des écarts entre la représentation de la réalité et la réalité peuvent nous mener à des résultats différents. Il y a ensuite l'enseignment de la physique statistique qui découle en partie de la première raison. Pour des éléments traités en très grand nombre, la réalisation du processus à l'ensemble des particules fait appel à la loi des grands nombres. C'est alors que des probabilités peuvent être introduites, même à nos échelles.

Pourquoi parler de physique alors que nous faisons de la robotique ?!?

La robotique est basée sur le monde physique et ne peut échapper à ses lois. Plus précisément, un robot a des capteurs soumis à des erreurs.

```python

import numpy as np

from slam_robot.models.world_items import Point, LineByTwoPoints, Circle
from slam_robot.models.robot import Robot
from slam_robot.models.world import World
from slam_robot.models.action import Turn, Sense, Move, Wait
initial_position = Point(12, 12)
initial_orientation = 0
robot = Robot(initial_position, initial_orientation)

actions = [
    Turn.from_objective(1.5, np.pi/3),
    Sense(),
    Move.from_objective(10, 10),
    Sense(),
    Turn.from_objective(1.5, np.pi/6),
    Move.from_objective(10, 30),
    Sense(),
    Move.from_objective(10, 30),
    Sense(),
    Wait(3)
]

world_1 = World(
    [
        LineByTwoPoints(Point(0, 0), Point(0, 100)),
        LineByTwoPoints(Point(0, 100), Point(100, 100)),
        LineByTwoPoints(Point(100, 100), Point(100, 0)),
        LineByTwoPoints(Point(100, 0), Point(0, 0)),
        Circle(Point(50, 50), 10),
    ],
    100,
    100
)


# trajectoire réelle du robot
trajectory = robot.recover_trajectory(actions, world_1, 0.5)
```

```python
trajectory[0].position, trajectory[0].orientation, trajectory[0].timestamp
```

```python
mean = 0
standard_deviation = 0.1
trajectory[0].position + Point(np.random.normal(mean, standard_deviation), np.random.normal(mean, standard_deviation))
```
